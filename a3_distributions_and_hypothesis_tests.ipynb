```python
import numpy as np
from scipy.stats import norm, chi2

# Question 1 (original answers were right, to rounding)
print(f'1.1 P(return < 0%)  = {norm.cdf(0, 14.7, 33):.1%}')
print(f'1.2 top-15% cutoff  = {norm.ppf(0.85, 14.7, 33):.1f}% annual return')
```

```python
# Question 2: one-tailed test, H0: mu = 20 vs Ha: mu < 20, sigma = 6, n = 20, sample mean = 14
mu0, mu1, sigma, n, xbar = 20, 16, 6, 20, 14
se = sigma / np.sqrt(n)
for alpha in (0.10, 0.05):
    z = (xbar - mu0) / se
    crit = mu0 - norm.ppf(1 - alpha) * se          # reject H0 if sample mean falls below this
    beta = 1 - norm.cdf((crit - mu1) / se)          # P(fail to reject | true mean is 16)
    print(f'alpha={alpha:.2f}: z={z:.2f}, p={norm.cdf(z):.1e} -> reject H0 | critical mean {crit:.2f} h | beta (miss a drop to 16 h) = {beta:.1%}')

z_a = z_b = norm.ppf(0.95)
n_needed = ((z_a + z_b) * sigma / (mu0 - mu1)) ** 2
print(); print(f'2.5 sample size for alpha = beta = 5%: n = {n_needed:.1f} -> {int(np.ceil(n_needed))} employees')
```

```python
# Question 3: OpenIntro 6.3.5, days until the S&P 500 has a positive day (1990-2011), geometric with p = 0.532
observed = np.array([1532, 760, 338, 194, 74, 33, 17])   # waits of 1, 2, 3, 4, 5, 6, 7+ days
N, p = observed.sum(), 0.532
probs = [(1 - p) ** (k - 1) * p for k in range(1, 7)]
probs.append(1 - sum(probs))                            # last bin is 7 or more
expected = N * np.array(probs)
stat = ((observed - expected) ** 2 / expected).sum()
df = len(observed) - 1
print('expected:', expected.round(0))
print(f'chi-square = {stat:.2f}, df = {df}, p-value = {chi2.sf(stat, df):.2f}')
```

**Conclusion:** the original reached the **right conclusion** (trading days are not independent), but the expected counts were computed by halving each observed count instead of using the geometric distribution, so the chi-square value (8.19) was off. With the correct geometric expectations, chi-square is about **15.2 on 6 df, p ≈ 0.02**. That is below 0.05, so we **reject** the geometric model: the waiting times between up days don't behave as they would if each day were independent. The biggest gaps are more 4-day waits and fewer 7+ day waits than independence predicts.

**Business impact:** risk and trading models that assume each day is an independent coin flip will misjudge how likely long losing or winning streaks are. Some short-term dependence exists, so streak behaviour should be modelled explicitly (and tested before anyone trades on it), rather than assumed away.
